### PySpark Assignment 2

1. Numbers

Create an RDD of [3, 6, 9, 12] with 2 partitions and divide every element by 3 (integer division).

2. Letter count

Given ["a b", "b c", "c a"], count how many times each letter appears.

3. Names

From ["Anu", "Ravi", "Asha", "Kiran"] keep only names starting with "A".

4. Big cities

Using sales, find cities whose total sales are greater than 300.

5. Grand total

Using sales, compute the sum of all sale amounts using map and reduce.

6. Classify

Which of these are transformations and which are actions: map, collect, filter, count, reduceByKey, first,
flatMap, take? Which causes a shuffle?

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder\
                    .appName('Assignment 2')\
                    .master('local[*]')\
                    .getOrCreate()
```

```python
sc = spark.sparkContext
```

#### Question 1

```python
rdd1 = sc.parallelize([3,6,9,12],2)
```

```python
rdd1.map(lambda x:int(x/3)).collect()
```

#### Question 2

```python
rdd2 = sc.parallelize(["a b", "b c", "c a"])
```

```python
rdd2.flatMap(lambda x:x.split(' ')).map(lambda x:(x,1)).reduceByKey(lambda x,y:x+y).collect()
```

#### Question 3

```python
rdd3 = sc.parallelize(["Anu", "Ravi", "Asha", "Kiran"])
```

```python
rdd3.filter(lambda x:x.startswith('A')).collect()
```

#### Question 4

```python
sales = sc.parallelize([("Chennai", 100), ("Delhi", 200), ("Chennai", 550), ("Bangalore",300),
                        ("Mumbai", 100), ("Delhi", 50), ("Mumbai", 120),("Bangalore",100)], 2)
```

```python
sales.reduceByKey(lambda x,y:x+y).filter(lambda x:x[1] > 300).collect()
```
